```python
import os
import pandas as pd
```

```python


def filter_csv_files(input_csv, directory_path):
    # Read the input CSV file
    df = pd.read_csv(input_csv)
    
    # Extract and split the names from the first column
    all_names = []
    for pair in df.iloc[:, 0]:
        names = pair.split('-')
        all_names.extend(names)
    
    # Get all CSV files in the target directory
    files_in_dir = os.listdir(directory_path)
    
    # Filter and delete files not in our list
    deleted_files = []
    kept_files = []
    for file in files_in_dir:
        if file.endswith('.csv'):
            # Extract the base name (without _1h.csv)
            base_name = file.rsplit('_', 1)[0]
            
            if base_name in all_names:
                kept_files.append(file)
            else:
                file_path = os.path.join(directory_path, file)
                os.remove(file_path)
                deleted_files.append(file)
    
    # Print summary
    print(f"Kept {len(kept_files)} files matching the names in the CSV.")
    print(f"Deleted {len(deleted_files)} files not in the CSV.")
    if deleted_files:
        print("Deleted files:")
        for f in deleted_files:
            print(f" - {f}")

# Example usage:
# if __name__ == "__main__":
#     input_csv = "your_input_file.csv"  # Replace with your CSV file path
#     directory_path = "your_directory"   # Replace with your target directory path
    
#     filter_csv_files(input_csv, directory_path)
```
